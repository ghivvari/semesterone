M3 S2

```python
n = int(input("Masukkan nilai n: "))
total = 0

for i in range(1, n + 1):
    if i % 2 == 1:
        total += i
    

print("Hasil penjumlahan:", total)
```

M4 S4, b

```python
n = (int(input("Masukkan n: ")))
i = 0
index = [0]

while i <= n:
    print(i)
    if i % 2 == 1:
        index.append(i)
    i+=1        

print(sum(index))    
```

M4 S4, c

```python
angka = int(input("Masukkan bilangan: "))
jumlah_digit = 0

if angka == 0:
    jumlah_digit = 1
else:
    while angka > 0:
        jumlah_digit += 1
        angka //= 10  

print("Jumlah digit:", jumlah_digit)
```

M5 S2

```python
numbers = [12, 75, 150, 180, 145, 525, 50]

for num in numbers:
    if num > 500:
        break
    if num > 150:
        continue
    if num % 5 == 0:
        print(num)
```
