```python
kelas = {
    "Anak 1": {
        "IPA": 90,
        "IPS": 80,
        "PJOK": 70,
    },
    "Anak 2": {
        "IPA": 75,
        "IPS": 65,
        "PJOK": 55,
    },
    "Anak 3": {
        "IPA": 85,
        "IPS": 75,
        "PJOK": 65,
    },
    "Anak 4": {
        "IPA": 60,
        "IPS": 50,
        "PJOK": 90
    }
}

for anak, score in kelas.items():
    if score["IPA"]> 70:
        if score["IPA"] > 80:
            print(f"{anak} qualify for olimpiade IPA with a score of {score['IPA']}")
        else:
            print(f"{anak} not qualify for olimpiade IPA with a score of {score['IPA']}")
    elif score["IPS"] > 70:
        if score["IPS"] > 80:
            print(f"{anak} qualify for olimpiade IPS with a score of {score['IPS']}")
        else:
            print(f"{anak} not qualify for olimpiade IPS with a score of {score['IPS']}")
    elif score["PJOK"] > 70:
        if score["PJOK"] > 80:
            print(f"{anak} qualify for olimpiade PJOK with a score of {score['PJOK']}")
        else:
            print(f"{anak} not qualify for olimpiade PJOK with a score of {score['PJOK']}")
    else:
        print(f"{anak} did not qualify for any olimpiade with scores IPA: {score['IPA']}, IPS: {score['IPS']}, PJOK: {score['PJOK']}")
```

```python
class LinearModel:
    def __init__(self, weight, bias):
        self.weight = weight
        self.bias = bias

    def predict(self, x):
        return self.weight * x + self.bias

my_model = LinearModel(weight= 3.0, bias= 2.0)

print(my_model.predict(3))  
```

```python
class LinearModel:
    def __init__(self, name, weight, bias):
        self.name = name
        self.weight = weight
        self.bias = bias
    
    def predoct(self,x):
        return self.weight + x + self.bias
    
A = LinearModel(name="A underfitted", weight= 3.0, bias= 3.0)
B = LinearModel(name="B balance", weight= 4.0, bias= 1.0)
C = LinearModel(name="C overfitted", weight= 5.0, bias= 6.0)

Akurasi = {
    A: 0.67,
    B: 0.99,
    C: 0.80
}

accuracy_threshold = 0.80
testinput = 3

for model_name, accuracy in Akurasi.items():
    if accuracy >= (accuracy_threshold + 0.10):
        print(f"\nmodel {model_name.name} passed with exceptional accuracy {accuracy * 100}")
        print(f"prediction {model_name.predoct(testinput)}")

    elif accuracy >= accuracy_threshold:
        print(f"\nmodel {model_name.name} passed with normal accuracy {accuracy * 100}")
        print(f"prediction {model_name.predoct(testinput)}")
    else:
        print(f"\nmodel {model_name.name} failed with poor accuracy {accuracy * 100}")
```

```python
class nambah_nambah:
    def __init__(self, a, b):
        self.a = b
        self.b = a
    def nambah(self):
        return self.a - self.b
print(nambah_nambah(4, 3).nambah())
```

```python
class ayam:
    def __init__(self, nama, jenis, berat):
        self.nama = str(nama)
        self.jenis = str(jenis)
        self.berat = int(berat)
    def pengenal(self):
        print(f"Nama ayam: {self.nama} \nJenis ayam: {self.jenis} \nBerat ayam: {self.berat}")
    def bb(self):
        if self.berat > 5:
            print(f"si {self.nama} berat bat jir")
        else:
            print(f"si {self.nama} cungkring bat")

ayam1 = ayam("prabowo", "cemani", 8)
ayam2 = ayam("gibran", "kampung", 5)

ayam1.pengenal()
ayam2.bb()
ayam1.bb()
```
