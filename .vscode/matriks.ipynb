```python
n = (int(input("Masukkan n: ")))
i = 0
index = [0]
while i <= n:
    index.append(i)
    i+=1

print(sum(index))    
```

```python
n = (int(input("Masukkan n: ")))
i = 0
index = [0]

while i <= n:
    print(i)
    if i % 2 == 1:
        index.append(i)
    i+=1        

print(sum(index))    
```

```python
n = int(input("Masukkan n: "))
total = 0

for i in range(n+1):
    print(i)
    if i % 2 == 1:
        total += i

print(total)
```

```python
x = int(input("Masukan angka: "))
while x != 2:
    x = int(input("Masukan angka: "))
else:
    print(f"anjay {x}")
```

```python
x = int(input("Masukan angka: "))
while x % 2 == 1:
    print(f"ganjil jir {x}")
    x = int(input("Masukan angka: "))
else:
    print(f"genap {x}")
```
